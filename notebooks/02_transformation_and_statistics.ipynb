# 2. Transformations and descriptive statistics

The transformed CSV is a classroom demonstration of one-hot encoding, min-max scaling, z-scores, and tool-wear quantile bins. It is derived across the complete descriptive dataset and must not be used to train or evaluate the failure model. Modeling uses the original 14 canonical columns, with preprocessing fitted inside Yahya's training pipeline.

```python
from pathlib import Path
import pandas as pd

from industrial_analyzer.data_loader import load_ai4i_csv
from industrial_analyzer.transformation import transform_for_dav, write_transformed_csv
from industrial_analyzer.statistics import (
    summary_statistics_table,
    failure_by_type_table,
    failure_modes_table,
    write_statistics_tables,
)

REPOSITORY_ROOT = Path.cwd()
assert (REPOSITORY_ROOT / "src/industrial_analyzer").is_dir(), "Start Jupyter from the repository root."
clean = load_ai4i_csv("data/processed/ai4i_cleaned.csv")
print(f"Canonical model/analytics input: {clean.shape}")
```

```python
transformed = transform_for_dav(clean)
write_transformed_csv(clean)
print(f"Demonstration transform: {transformed.shape}")
print("Saved: data/processed/ai4i_transformed.csv")
display(transformed.head())
```

```python
tables = {
    "Summary statistics": summary_statistics_table(clean),
    "Failure by product type": failure_by_type_table(clean),
    "Recorded failure modes": failure_modes_table(clean),
}
written = write_statistics_tables(clean)
for title, table in tables.items():
    print(title)
    display(table)
print("Saved table files:")
for name, path in written.items():
    print(f"{name}: {path.relative_to(REPOSITORY_ROOT).as_posix()}")
```

Failure-mode flags may overlap. These counts describe labels in the synthetic dataset; they do not establish causes or equipment safety limits.
